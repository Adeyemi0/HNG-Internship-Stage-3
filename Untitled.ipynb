```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage

import warnings
warnings.filterwarnings("ignore")
```

```python
# Read the Excel file into a DataFrame
df = pd.read_csv('data.csv')

# Display the first few rows of the DataFrame
df.head()
```

## Standardizing the columns and creating key performance metrics

```python
# Check if every entry ends with 'days'
all_in_days = df['Duration'].str.endswith('days').all()

print("All durations in days:", all_in_days)
```

```python
# Extract the numeric part of the duration
df['Duration'] = df['Duration'].str.extract('(\d+)').astype(int)
```

```python
# Check if every entry starts with '$'
all_start_with_dollar = df['Acquisition_Cost'].astype(str).str.startswith('$').all()
print("All Acquisition Costs start with '$':", all_start_with_dollar)
```

```python
# Extract the numeric part of the Acquisition_Cost
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace({'\$': '', ',': ''}, regex=True).astype(float)
```

```python
df['Conversions'] = df['Clicks'] * df['Conversion_Rate']
```

```python
df['CPC'] = df['Acquisition_Cost'] / df['Clicks']
```

```python
df['Engagement_Rate'] = df['Engagement_Score'] / df['Impressions']
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
df['CTR'] = (df['Clicks'] / df['Impressions']) * 100
```

```python
df['Cost_per_Conversion'] = df['Acquisition_Cost'] / df['Conversions']
```

```python
# checking the datatypes
df.dtypes
```

## Summary Statistics

```python
# Summary Statistics for Numerical Columns
numerical_columns = ['Conversion_Rate', 'Acquisition_Cost', 'ROI', 'Clicks', 'Impressions', 
                     'Engagement_Score', 'Conversions', 'CPC', 'Engagement_Rate', 'CTR', 'Cost_per_Conversion']

# Calculate summary statistics
summary_stats = df[numerical_columns].describe().T
print("Summary Statistics:")
summary_stats
```

## Checking for outliers

```python
columns_of_interest = ['Conversion_Rate', 'Acquisition_Cost', 'ROI', 'Clicks', 'Impressions', 
                     'Engagement_Score']

# Identify outliers using IQR method
outliers = {}
for col in columns_of_interest:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers[col] = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col].tolist()

print("\nOutliers by Column:")
for col, outlier_values in outliers.items():
    if len(outlier_values) > 0:
        print(f"{col}: {outlier_values}")
    else:
        print(f"{col}: No outliers detected")
```

## Frequency Tables for Categorical Variables

```python
# Frequency Tables for Categorical Variables
categorical_columns = ['Company', 'Campaign_Type', 'Target_Audience', 'Channel_Used', 'Location', 'Customer_Segment']

# Create frequency tables
frequency_tables = {}
for col in categorical_columns:
    freq_table = df[col].value_counts(normalize=True).head(5)  # Top 5 most frequent categories
    frequency_tables[col] = freq_table
```

```python
print("\nFrequency Tables:")
for col, freq_table in frequency_tables.items():
    print(f"\n{col} Frequencies:")
    print(freq_table)
```

## Correlation Matrix Table

```python
# Correlation Matrix
numerical_columns = ['Conversion_Rate', 'Acquisition_Cost', 'ROI', 'Clicks', 'Impressions', 
                     'Engagement_Score', 'Conversions', 'CPC', 'Engagement_Rate', 'CTR', 'Cost_per_Conversion']

# Compute pairwise correlations
correlation_matrix = df[numerical_columns].corr()

# Display the correlation matrix
print("Correlation Matrix:")
correlation_matrix
```

## Correlation Matrix with a Heatmap

```python
# Visualize the Correlation Matrix with a Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title('Correlation Matrix Heatmap')
plt.savefig('Correlation_Matrix_Heatmap.png')
plt.show()
```

```python
# Define a threshold for strong correlations 
threshold = 0.5

# Extract pairs with strong correlations
strong_correlations = []
for col1 in correlation_matrix.columns:
    for col2 in correlation_matrix.index:
        if col1 != col2:  # Exclude self-correlations
            correlation_value = correlation_matrix.loc[col1, col2]
            if abs(correlation_value) > threshold:
                strong_correlations.append((col1, col2, correlation_value))

# Print strong correlations
print("\nStrong Correlations (|r| > 0.5):")
for col1, col2, corr_val in strong_correlations:
    print(f"{col1} vs {col2}: {corr_val:.2f}")
```

```python
# Is higher CTR associated with better Conversion_Rate?
ctr_conversion_corr = correlation_matrix.loc['CTR', 'Conversion_Rate']
print(f"\nCorrelation between CTR and Conversion_Rate: {ctr_conversion_corr:.2f}")

# Does CPC negatively impact ROI?
cpc_roi_corr = correlation_matrix.loc['CPC', 'ROI']
print(f"Correlation between CPC and ROI: {cpc_roi_corr:.2f}")

# Does Engagement_Score lead to more Conversions?
engagement_conversions_corr = correlation_matrix.loc['Engagement_Score', 'Conversions']
print(f"Correlation between Engagement_Score and Conversions: {engagement_conversions_corr:.2f}")
```

## SEGMENT ANALYSIS

```python
# Group data by Target_Audience, Customer_Segment, or Location
segments = ['Target_Audience', 'Customer_Segment', 'Location']
kpi_columns = ['Conversion_Rate', 'ROI', 'CTR', 'CPC', 'Cost_per_Conversion']

# Function to calculate average KPIs for each segment
def analyze_segment_performance(segment_col):
    segment_analysis = df.groupby(segment_col)[kpi_columns].mean().reset_index()
    return segment_analysis

# Analyze each segment
segment_results = {}
for segment in segments:
    segment_results[segment] = analyze_segment_performance(segment)
```

```python
# Which Customer_Segment has the highest Conversion_Rate?
highest_conversion_segment = segment_results['Customer_Segment'].loc[
    segment_results['Customer_Segment']['Conversion_Rate'].idxmax()]
print(f"\nCustomer_Segment with Highest Conversion_Rate:\n{highest_conversion_segment}")
```

```python
# Visualize Conversion_Rate by Customer_Segment
plt.figure(figsize=(10, 6))
sns.barplot(data=segment_results['Customer_Segment'], x='Customer_Segment', y='Conversion_Rate')
plt.title('Average Conversion Rate by Customer Segment')
plt.xticks(rotation=45)
plt.savefig('Average_Conversion_Rate_by_Customer_Segment.png')
plt.show()
```

# ANOVA ANALYSIS (Campaign Type)

```python
import pandas as pd
import numpy as np
from scipy.stats import f_oneway
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Assuming 'df' is your DataFrame

# 1. Perform ANOVA for Campaign_Type
campaign_types = df['Campaign_Type'].unique()
roi_by_campaign_type = [df[df['Campaign_Type'] == ctype]['ROI'].dropna() for ctype in campaign_types]

# Check if there are enough data points for each group
if any(len(group) < 2 for group in roi_by_campaign_type):
    print("Not enough data for some Campaign_Type groups to perform ANOVA.")
else:
    # Perform one-way ANOVA
    f_stat, p_value = f_oneway(*roi_by_campaign_type)
    print(f"ANOVA Results for Campaign_Type:")
    print(f"F-statistic: {f_stat:.4f}")
    print(f"P-value: {p_value:.4f}")

    # Interpretation
    alpha = 0.05  # Significance level
    if p_value < alpha:
        print("There is a statistically significant difference in ROI between Campaign_Types.")
    else:
        print("There is no statistically significant difference in ROI between Campaign_Types.")
```

# ANOVA ANALYSIS (Channel Used)

```python
# Perform ANOVA for Channel_Used
channels = df['Channel_Used'].unique()
roi_by_channel_used = [df[df['Channel_Used'] == channel]['ROI'].dropna() for channel in channels]

# Check if there are enough data points for each group
if any(len(group) < 2 for group in roi_by_channel_used):
    print("\nNot enough data for some Channel_Used groups to perform ANOVA.")
else:
    # Perform one-way ANOVA
    f_stat, p_value = f_oneway(*roi_by_channel_used)
    print(f"\nANOVA Results for Channel_Used:")
    print(f"F-statistic: {f_stat:.4f}")
    print(f"P-value: {p_value:.4f}")

    # Interpretation
    alpha = 0.05  # Significance level
    if p_value < alpha:
        print("There is a statistically significant difference in ROI between Channels.")
    else:
        print("There is no statistically significant difference in ROI between Channels.")
```
